```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier #create a classifier
from sklearn.metrics import classification_report

#TAHAPAN 1 IMPORT DATA
df = pd.read_csv('Iris.csv')
df.head()

#melihat missing value (nilai yang kosong)
df.isnull().sum()

#memisahkan variabel dan class
X = df.drop(['Id', 'Species'], axis=1)
y = df[['Species']]
X.head()

# SPLIT DATA TRAINING = 70 %, DATA TESTING = 30 %
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size = 0.3, random_state=0)
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)

# TAHAPAN 3 ADALAH TRAINING MODEL
model = RandomForestClassifier(n_estimators=40)
#train the model
modelfit = model.fit(X_train, y_train)
accuracy = modelfit.score(X_test, y_test)
#predict the response
y_pred = model.predict(X_test)
print(accuracy)

#TAHAPAN 4 ADALAH EVALUASI MODEL DENGAN CONFUSION MATRIC 
print(classification_report(y_test, y_pred))

# Import model ke File pickle 
import pickle
with open('RF_pickle','wb') as r:
   pickle.dump(modelfit,r)
```
